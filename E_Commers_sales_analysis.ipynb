```python
import pandas as pd
import numpy as np

df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

print("Dataset loaded successfully!")
print("Rows and Columns:", df.shape)

display(df.head())
```

```python
print("First 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("\nRandom 5 Rows:")
display(df.sample(5))

print("\nShape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
print("Dataset Information:")
df.info()

print("\nDescriptive Statistics:")
display(df.describe())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sale = df["Total_Sales"].max()
lowest_sale = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print("Total Sales:", total_sales)
print("Average Sale:", average_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
print("Total Quantity Sold:", total_quantity)
print("Number of Orders:", number_of_orders)
```

```python
category_sales = (
    df.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Category-wise Sales:")
display(category_sales)
```

```python
city_sales = (
    df.groupby("City")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("City-wise Sales:")
display(city_sales)
```

```python
product_sales = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Product-wise Sales:")
display(product_sales)
```

```python
payment_sales = (
    df.groupby("Payment_Method")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Payment Method-wise Sales:")
display(payment_sales)
```

```python
region_sales = (
    df.groupby("Region")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Region-wise Sales:")
display(region_sales)
```

```python
print("Top 5 Products by Sales:")
display(product_sales.head(5))
```

```python
print("Bottom 5 Products by Sales:")
display(product_sales.tail(5))
```

```python
print("Top 10 Orders by Sales:")

top_orders = df.sort_values(
    "Total_Sales",
    ascending=False
).head(10)

display(top_orders)
```

```python
print("Orders with Total Sales Greater Than 10000:")

high_sales_orders = df[df["Total_Sales"] > 10000]

display(high_sales_orders)
```

```python
top_category = category_sales.idxmax()
top_city = city_sales.idxmax()
top_product = product_sales.idxmax()
top_region = region_sales.idxmax()
top_payment = payment_sales.idxmax()

print("===== FINAL OBSERVATIONS =====")

print("1. Highest-selling category:", top_category)
print("2. Highest-selling city:", top_city)
print("3. Highest-selling product:", top_product)
print("4. Highest-selling region:", top_region)
print("5. Highest-sales payment method:", top_payment)
print("6. Total sales:", total_sales)
print("7. Total quantity sold:", total_quantity)
print("8. Total number of orders:", number_of_orders)
```

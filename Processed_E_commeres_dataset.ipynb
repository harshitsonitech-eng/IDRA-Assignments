```python
import pandas as pd
import glob
import os

print("Available CSV files:")

for file in glob.glob("*.csv"):
    print(file)
```

```python
orders_file = "Day9_Orders.csv"
customers_file = "Day9_Customers.csv"

product_files = glob.glob("Day9_Products*.csv")

if len(product_files) == 0:
    raise FileNotFoundError("Day9_Products CSV nahi mili.")

products_file = product_files[0]

orders = pd.read_csv(orders_file)
customers = pd.read_csv(customers_file)
products = pd.read_csv(products_file)

orders.columns = orders.columns.str.strip()
customers.columns = customers.columns.str.strip()
products.columns = products.columns.str.strip()

print("Orders:", orders.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)

print("\nOrders Columns:")
print(orders.columns.tolist())

print("\nCustomers Columns:")
print(customers.columns.tolist())

print("\nProducts Columns:")
print(products.columns.tolist())
```

```python
print("ORDERS")
display(orders.head())

print("\nCUSTOMERS")
display(customers.head())

print("\nPRODUCTS")
display(products.head())
```

```python
print("Orders Missing Values:")
print(orders.isnull().sum())

print("\nCustomers Missing Values:")
print(customers.isnull().sum())

print("\nProducts Missing Values:")
print(products.isnull().sum())

print("\nOrders Data Types:")
print(orders.dtypes)

print("\nCustomers Data Types:")
print(customers.dtypes)

print("\nProducts Data Types:")
print(products.dtypes)
```

```python
merged_data = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)

print("After Orders + Customers merge:")
print(merged_data.shape)

display(merged_data.head())
```

```python
merged_data = pd.merge(
    merged_data,
    products,
    on="Product_ID",
    how="left",
    suffixes=("", "_Product")
)

print("Final merged shape:")
print(merged_data.shape)

display(merged_data.head())
```

```python
part1 = merged_data.iloc[:10]
part2 = merged_data.iloc[10:]

processed_data = pd.concat(
    [part1, part2],
    ignore_index=True
)

print("Data after concat():")
print(processed_data.shape)

display(processed_data.head())
```

```python
processed_data["Order_Date"] = pd.to_datetime(
    processed_data["Order_Date"],
    dayfirst=True,
    errors="coerce"
)

processed_data["Order_Month"] = processed_data["Order_Date"].dt.month
processed_data["Order_Day"] = processed_data["Order_Date"].dt.day
processed_data["Order_Day_Name"] = processed_data["Order_Date"].dt.day_name()

print("DateTime operations completed.")

display(
    processed_data[
        ["Order_Date", "Order_Month", "Order_Day", "Order_Day_Name"]
    ].head()
)
```

```python
processed_data["Quantity"] = pd.to_numeric(
    processed_data["Quantity"],
    errors="coerce"
)

processed_data["Unit_Price"] = pd.to_numeric(
    processed_data["Unit_Price"],
    errors="coerce"
)

processed_data["Customer_Name_Clean"] = processed_data["Customer_Name"].apply(
    lambda x: str(x).strip().title()
)

processed_data["Calculated_Total"] = (
    processed_data["Quantity"] * processed_data["Unit_Price"]
)

print("apply() and calculated columns completed.")

display(
    processed_data[
        [
            "Customer_Name",
            "Customer_Name_Clean",
            "Quantity",
            "Unit_Price",
            "Calculated_Total"
        ]
    ].head()
)
```

```python
print("Final Processed Dataset")
print("Rows:", processed_data.shape[0])
print("Columns:", processed_data.shape[1])

print("\nColumns:")
print(processed_data.columns.tolist())

print("\nMissing Values:")
print(processed_data.isnull().sum())

display(processed_data.head(10))
```

```python
print("Sales by Category:")

category_sales = processed_data.groupby(
    "Category"
)["Calculated_Total"].sum().sort_values(ascending=False)

display(category_sales)
```

```python
print("Top Performing Products:")

product_sales = processed_data.groupby(
    ["Product_ID", "Product_Name"]
)["Calculated_Total"].sum().sort_values(ascending=False)

display(product_sales.head(10))
```

```python
print("Order Status Summary:")

status_summary = processed_data["Order_Status"].value_counts()

display(status_summary)
```

```python
output_file = "Day9_Processed_Ecommerce_Dataset.csv"

processed_data.to_csv(
    output_file,
    index=False
)

print("SUCCESS! Final processed dataset exported as:")
print(output_file)

print("\nFile size:", os.path.getsize(output_file), "bytes")
```

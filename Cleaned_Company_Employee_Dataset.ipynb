```python
import pandas as pd
import numpy as np
import glob
import os

from google.colab import files

uploaded = files.upload()

print("Uploaded files:")
for f in uploaded:
    print(f)
```

```python
csv_files = glob.glob("*.csv")

if len(csv_files) == 0:
    raise FileNotFoundError("CSV file not found.")

file_name = csv_files[0]
df = pd.read_csv(file_name)

df.columns = df.columns.str.strip()

print("File loaded:", file_name)
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

display(df.head())
```

```python
print("Dataset Information:")
df.info()

print("\nShape:", df.shape)

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing Values Before Cleaning:")

missing_values = df.isnull().sum()

display(missing_values[missing_values > 0])

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
categorical_columns = [
    "Department",
    "Job_Title",
    "Gender",
    "City",
    "Work_Mode"
]

for col in categorical_columns:
    print("\n", col)
    print(df[col].astype(str).str.strip().value_counts().head(20))
```

```python
numeric_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce",
    dayfirst=True
)

print("Data types corrected.")

print(df.dtypes)
```

```python
numeric_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

print("Numeric missing values filled using median.")
print(df[numeric_columns].isnull().sum())
```

```python
categorical_columns = [
    "Department",
    "Job_Title",
    "Gender",
    "City",
    "Work_Mode"
]

for col in categorical_columns:
    if df[col].isnull().any():
        mode_value = df[col].mode()

        if len(mode_value) > 0:
            df[col] = df[col].fillna(mode_value[0])

print("Categorical missing values filled using mode.")
print(df[categorical_columns].isnull().sum())
```

```python
df["Joining_Date"] = df["Joining_Date"].ffill()

print("Joining_Date missing values after forward fill:")
print(df["Joining_Date"].isnull().sum())
```

```python
text_columns = [
    "Employee_ID",
    "Employee_Name",
    "Department",
    "Job_Title",
    "Gender",
    "City",
    "Work_Mode"
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

df["Gender"] = df["Gender"].replace({
    "M": "Male",
    "m": "Male",
    "F": "Female",
    "f": "Female",
    "N/A": "Other",
    "NA": "Other"
})

df["Work_Mode"] = df["Work_Mode"].replace({
    "office": "Office",
    "OFFICE": "Office",
    "remote": "Remote",
    "REMOTE": "Remote",
    "hybrid": "Hybrid",
    "HYBRID": "Hybrid"
})

print("Text values cleaned and standardized.")
```

```python
before_duplicates = len(df)

df = df.drop_duplicates()

after_duplicates = len(df)

print("Rows before removing duplicates:", before_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

```python
print("----- FINAL CLEANING CHECK -----")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Rows:")
print(df.duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)

display(df.head(10))
```

```python
print("----- BEFORE vs AFTER CLEANING -----")

print("Original rows:", len(pd.read_csv(file_name)))
print("Cleaned rows:", len(df))

print("\nOriginal columns:", len(pd.read_csv(file_name).columns))
print("Cleaned columns:", len(df.columns))

print("\nRemaining missing values:", df.isnull().sum().sum())
print("Remaining duplicate rows:", df.duplicated().sum())
```

```python
print("""
CLEANING SUMMARY

1. Loaded the company employee dataset using Pandas.
2. Inspected dataset shape, columns and data types.
3. Identified missing values.
4. Converted numeric columns to appropriate numeric data types.
5. Converted Joining_Date into datetime format.
6. Filled numeric missing values using median.
7. Filled categorical missing values using mode.
8. Used forward filling for missing joining dates.
9. Removed extra spaces from text values.
10. Standardized inconsistent categorical entries.
11. Removed duplicate records.
12. Verified the dataset after cleaning.
""")
```

```python
output_file = "Day11_Cleaned_Company_Employee_Dataset.csv"

df.to_csv(
    output_file,
    index=False
)

print("SUCCESS! Cleaned dataset exported successfully.")
print("File name:", output_file)
print("File size:", os.path.getsize(output_file), "bytes")
```
